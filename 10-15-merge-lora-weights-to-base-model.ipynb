# 將Lora合併到基礎模型中

```python
from peft import PeftModel
from transformers import AutoModelForCausalLM, AutoTokenizer, AutoConfig
import torch
```

```python
model_name_or_path = 'YeungNLP/bloomz-396m-zh'
```

```python

config = AutoConfig.from_pretrained(model_name_or_path)
tokenizer = AutoTokenizer.from_pretrained(
    model_name_or_path,
    trust_remote_code=True,
    # llama不支持fast
    use_fast=True
)
```

```python
# 加载模型

device_map='auto'
model = AutoModelForCausalLM.from_pretrained(
    model_name_or_path,
    trust_remote_code=True,
    low_cpu_mem_usage=True,
    torch_dtype=torch.float16,
    device_map='auto',
    #device_map={'': 'cpu'}
)
```

```python

adapter_name_or_path = 'my-checkpoints/checkpoint-640'
model = PeftModel.from_pretrained(model, adapter_name_or_path, device_map=device_map)
```

```python
model = model.merge_and_unload()
```

```python
save_path = 'my-lora-merged-model-steps640-YeungNLP'
tokenizer.save_pretrained(save_path)
model.save_pretrained(save_path)
```
